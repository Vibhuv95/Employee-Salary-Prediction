### Salary Distribution

```python
plt.figure(figsize=(8,5))
sns.histplot(df['salary'], bins=25, kde=True)
plt.title("Salary Distribution")
plt.show()
```

Employee Salaries are concentrated around 140k to 150k.
The salary distribution is slightly right-skewed, means that a small no. of employees earn substantially higher salary than the majority.
Most Employee earn b/t 120k to 170k.

```python
plt.figure(figsize=(8,5))
sns.boxplot(x=df["salary"])
plt.title("Salary Distribution")
```

Most Salaries are concentrated within the interquartile range, while a small no. of employees receive excaptionally high salaries.
A few unusually low salary values are also present.

# Salary Prediction Graphs 
### Comparing all other parameters with respect to salary

```python
sns.scatterplot(x='experience_years',y='salary',data=df)
```

Salary shows a moderate positive relationship with years of experience. 
As experience increases, both the minimum and maximum salary levels tend to increase, indicating that experienced employees generally earn higher salaries.

```python
sns.scatterplot(x='skills_count',y='salary',data=df)
```

Employees with a higher number of skills tend to earn slightly higher salaries. 
However, the relationship is relatively weak compared with years of experience.

```python
sns.boxplot(x='certifications',y='salary',data=df)
```

Employees with more certifications generally have higher median salaries. 
The salary distribution shifts upward as the number of certifications increases, suggesting certifications have a positive effect on salary.

```python
sns.boxplot(x='education_level',y='salary',data=df)
```

Employees with PhD and Master's degrees have higher median salaries than those with Bachelor's, Diploma, or High School education. This suggests that higher educational qualifications are associated with higher salaries.

```python
sns.boxplot(x='job_title',y='salary',data=df)
plt.xticks(rotation=45)
```

AI Engineers, Backend Developers, and Product Managers have higher median salaries than many other job roles. 
Data Analysts and Business Analysts show comparatively lower median salaries in this dataset. 
alary distributions vary noticeably across job titles, indicating that job role is an important predictor of salary.

```python
sns.boxplot(x='company_size',y='salary',data=df)
```

Enterprise companies offer the highest median salaries, followed by large companies. 
Startup companies have the lowest median salaries. 
This suggests that company size has a positive influence on employee salaries.

```python
sns.boxplot(x='industry',y='salary',data=df)
plt.xticks(rotation=45)
```

Salary distributions are relatively similar across industries. 
No industry appears to consistently offer significantly higher or lower salaries than the others.
